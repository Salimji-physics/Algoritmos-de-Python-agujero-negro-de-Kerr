```python
%display latex
M = Manifold(4, 'M') #Variedad diferenciable 4-dimensional
X.<t, r, th, ph> = M.chart(r't r th:\theta ph:\phi') #Carta en coordenadas de Boyer-Lindquist
var('m, a')#Constantes (masa y velocidad angular)

#Variables agrupadas
rho2, Delta = r^2 + (a*cos(th))^2, r^2 - 2*m*r + a^2

#Definición de la métrica de Kerr
g = M.metric('g')
g[0,0], g[0,3], g[1,1], g[2,2], g[3,3] = -(1 - 2*m*r/rho2), -2*a*m*r*sin(th)^2 / rho2, rho2 / Delta, rho2, (r^2 + a^2 + 2*m*r*(a*sin(th))^2 / rho2) * sin(th)^2

g.display()
```

```python
#Cálculo de la curvatura de Riemann y de Ricci
R, Ric=g.riemann(), g.ricci()
```

```python
#Inversión de la métrica
g_inv = g.inverse()
```

```python
R_cov=R.down(g) #Bajada de índices a un R covariante
R_cont=R.up(g) #Subida de índices a un R contravariante
```

```python
#Escalar de Kretschmann (invariante de Riemann)
K=(R_cov['_abcd']*R_cont['^abcd']).expr().simplify_rational().factor() #Suma de Einstein, simplificación de la expresión algebraica y factorización
```
